# 07 - Validade e Inferência Lógica na Segurança
Demonstração de inferência dedutiva (*Forward Chaining*, *Modus Ponens* e *Modus Tollens*) para validação de regras do Sistema Intertravado de Segurança (SIS).

```python
from typing import Dict, List, Set

# 1. Base de Conhecimento (KB) - Regras de Segurança do SIS
regras_seguranca = [
    {"id": "R1", "premissas": ["PT101_ALTO"], "conclusao": "TRIP_ESD100", "desc": "Se Pressão Alta -> Trip ESD"},
    {"id": "R2", "premissas": ["TT101_ALTO"], "conclusao": "TRIP_ESD100", "desc": "Se Temp Alta -> Trip ESD"},
    {"id": "R3", "premissas": ["TRIP_ESD100"], "conclusao": "FECHAR_XV101", "desc": "Se Trip ESD -> Fechar Válvula"},
    {"id": "R4", "premissas": ["TRIP_ESD100"], "conclusao": "LIGAR_ALARME", "desc": "Se Trip ESD -> Alarme Sonoro"}
]

# 2. Motor de Inferência por Encadeamento para Frente (Forward Chaining)
def motor_inferencia(fatos_iniciais: Set[str], regras: List[Dict]) -> Set[str]:
    fatos = set(fatos_iniciais)
    novos_fatos = True
    while novos_fatos:
        novos_fatos = False
        for regra in regras:
            if regra["conclusao"] not in fatos:
                if all(p in fatos for p in regra["premissas"]):
                    fatos.add(regra["conclusao"])
                    novos_fatos = True
    return fatos

# 3. Regras Formalizadas de Validação Dedução Lógica
def verificar_modus_ponens(p: bool, p_implica_q: bool) -> bool:
    return p and p_implica_q

def verificar_modus_tollens(nao_q: bool, p_implica_q: bool) -> bool:
    return nao_q and p_implica_q

# 4. Avaliação dos Cenários Operacionais
cenarios = {
    "Cenário 1: Pressão Alta (PT-101 > 5.0 bar)": {"PT101_ALTO"},
    "Cenário 2: Temperatura Alta (TT-101 > 85°C)": {"TT101_ALTO"},
    "Cenário 3: Operação Normal": set()
}

print(f"{'Cenário':<42} | {'Fatos Iniciais':<18} | {'Conclusões Inferidas':<35}")
print("-" * 100)

for nome, fatos in cenarios.items():
    conclusoes = motor_inferencia(fatos, regras_seguranca)
    novas_conclusoes = sorted(list(conclusoes - fatos))
    str_fatos = ", ".join(sorted(list(fatos))) if fatos else "Nenhum"
    str_conclusoes = ", ".join(novas_conclusoes) if novas_conclusoes else "Nenhuma (Normal)"
    print(f"{nome:<42} | {str_fatos:<18} | {str_conclusoes:<35}")

print("\n--- VALIDAÇÃO DE REGRAS DE INFERÊNCIA DEDUTIVA ---")
print(f"Modus Ponens  [PT-101 Alto => Trip ESD-100]: VÁLIDO? {verificar_modus_ponens(True, True)}")
print(f"Modus Tollens [Sem Trip => Pressão Normal] : VÁLIDO? {verificar_modus_tollens(True, True)}")
```
